```python
def prepare_X(df):
    """
    Prepare feature matrix X by handling numerical and categorical variables
    
    Args:
        df: Input DataFrame containing car features
        
    Returns:
        numpy array with processed features
    """
    # Create a copy of the dataframe to avoid modifying the original
    df_copy = df.copy()
    
    # Numerical features
    numeric_features = ['engine_hp', 'engine_cylinders', 'highway_mpg', 
                       'city_mpg', 'popularity']
    
    # Calculate age feature
    df_copy['age'] = 2017 - df_copy['year']
    features = numeric_features + ['age']

    # Handle doors - create dummy variables for different number of doors
    for num_doors in [2, 3, 4]:
        col_name = f'num_doors_{num_doors}'
        df_copy[col_name] = (df_copy.number_of_doors == num_doors).astype('int')
        features.append(col_name)

    # Handle categorical variables
    categorical_vars = {
        'make': df_copy.make.value_counts().head().index,
        'engine_fuel_type': df_copy.engine_fuel_type.value_counts().head().index,
        'transmission_type': df_copy.transmission_type.value_counts().head().index,
        'driven_wheels': df_copy.driven_wheels.value_counts().head().index,
        'market_category': df_copy.market_category.value_counts().head().index,
        'vehicle_size': df_copy.vehicle_size.value_counts().head().index,
        'vehicle_style': df_copy.vehicle_style.value_counts().head().index
    }

    # Create dummy variables for each category
    for category, values in categorical_vars.items():
        for val in values:
            col_name = f'{category}_{val}'
            df_copy[col_name] = (df_copy[category] == val).astype('int')
            features.append(col_name)

    # Fill missing values with 0 and select only the engineered features
    X = df_copy[features].fillna(0).values
    return X
```

```python
#regularization
```

```python
def train_linear_regression_regularized(X,y,r = 0.001):
    ones = np.ones((X.shape[0]))
    X = np.column_stack((ones, X))

    XTX = X.T.dot(X)
    XTX = XTX + r*np.eye(XTX.shape[0])


    XTX_inv = np.linalg.inv(XTX)
    w_full = XTX_inv.dot(X.T).dot(y)
    
    
    w0 = w_full[0]
    w = w_full[1:]
    
    
    return w0, w
```

```python
X_train = prepare_X(df_train)
w0, w = train_linear_regression_regularized(X_train, y_train, r = 0.01)

X_val = prepare_X(df_val)
y_pred = w0 + X_val.dot(w)

rmse(y_val, y_pred)
```

```python
sns.histplot(y_pred, color = 'red', alpha = 0.5, bins=50)
sns.histplot(y_val, color = 'blue', alpha=0.5, bins=50)
```

```python
# tuning the model

for r in [0.0, 0.0001, 0.001, 0.01, 0.1, 1.0, 10.0]:
    X_train = prepare_X(df_train)
    w0, w = train_linear_regression_regularized(X_train, y_train, r = r)

    X_val = prepare_X(df_val)
    y_pred = w0 + X_val.dot(w)

    score = rmse(y_val, y_pred)
    print('RMSE (r=%.4f): %.4f' % (r, score))
```
